```python
productos = ["Hamburguesa", "Pizza", "Papas fritas", "Gaseosa"]
precios = [8000, 10000, 4500, 2500]
pedido_nombres = []
pedido_precios = []
pedido_cantidades = []

print("¡Bienvenido al Roti AMADEUS!")

pidiendo = True
while pidiendo == True:
    print("--- CARTA ---")
    for i in range(len(productos)):
        print(f"{i + 1}. {productos[i]} - ${precios[i]}")

    opcion = int(input("Elige el número del producto (o escribe 0 para terminar): "))

    if opcion == 0:
        pidiendo = False
    elif opcion >= 1 and opcion <= len(productos):
        indice = opcion - 1
        cantidad = int(input(f"¿Cuántas unidades de {productos[indice]} quieres?: "))
        pedido_nombres.append(productos[indice])
        pedido_precios.append(precios[indice])
        pedido_cantidades.append(cantidad)
        print(f"Agregaste: {cantidad} x {productos[indice]}")
    else:
        print("Opción no válida. Intenta de nuevo.")

print("--- RESUMEN DE TU PEDIDO ---")
total = 0

for i in range(len(pedido_nombres)):
    subtotal = pedido_precios[i] * pedido_cantidades[i]
    print(f"- {pedido_cantidades[i]} x {pedido_nombres[i]}: $ {subtotal}")
    total = total + subtotal

print(f"Total a pagar: ${total}")
print("¡Gracias por tu compra!")
```
